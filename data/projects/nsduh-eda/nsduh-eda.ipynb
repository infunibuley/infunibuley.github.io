## Initial Data Cleaning

Checking for all columns with only 1 unique value 

```python
# get num unique values per column
unique_val_per_col = {col: nsduh_df[col].nunique() for col in nsduh_df.columns}
unique_val_per_col
```

```python
for col in unique_val_per_col:
    if unique_val_per_col[col] == 1:
        print(col + " : " + nsduh_df[col].iloc[0])
```

```python
nsduh_df.drop(columns='FILEDATE', inplace=True)
```

```python
del unique_val_per_col
```

Check NaN values

```python
nan_per_col = {col: nsduh_df[col].isna().sum() for col in nsduh_df.columns}
nan_per_col
```

```python
for col in nan_per_col:
    if nan_per_col[col] != 0:
        print(col + " : " + str(nan_per_col[col]))
```

#### 1A. What is the gap between needing mental health care versus actually receiving it?

#### 1B. How do these unmet needs and treatment rates differ across various demographic groups (race, income, gender, and age brackets)?

#### 1C. Are marginalized communities and specific cultural groups experiencing higher barriers to accessing treatment?

### 2. Disparities in Marginalized & Minority Communities

#### 2A. Intersectionality and Minority Stress: How do sample sizes for different respondent groups allow us to examine unique mental health hurdles faced by sexual and gender minority (SGM) populations compared to general population baselines?

#### 2B. Are specific racial/ethnic groups able to access affirming mental health support at the same rate, or are there massive systemic barriers in treatment?

#### 2C. Cultural Stigma & Models of Distress: How do cultural frameworks (such as the internalization of the model minority myth among Asian Americans) impact mental health reporting and attitudes toward seeking professional help?

### 3. Gender Dynamics (Focusing on Men's vs Women's Mental Health)

#### 3A. Are there any general differences between mental health concerns in women compared to men? 

#### 3B. How do hegemonic norms of masculinity and patriarchal structures influence men's willingness to seek mental health guidance or report psychological distress?

#### 3C. Are men experiencing a significantly wider help-seeking gap compared to women across similar socioeconomic and demographic brackets?

### 4. Generational Differences & Age-Related Trends

#### 4A. Help-Seeking Proactivity: Are younger generations (e.g., Gen Z and Millennials) more proactive about engaging in therapy and mental health care compared to older generations?

#### 4B. Substance Use vs. Abuse Patterns: What are the generational differences in substance use and dependence behaviors?

#### 4C. Socioeconomic Stressors: What are the correlations between socioeconomic/financial stressors and the use of substances as a coping mechanism across different age groups?
