PDEN10

The population density is based on 2010 Census Data. 1 is for 1mil pop-den or more, 2 is for less than 1mil pop-den, 3 is for unknown. 

```python
cleaned_df['PDEN10'].value_counts()
```

```python
PDEN10_mapping = {
    1: "High",
    2: "Low",
    3: "Unknown"
}
cleaned_df["PDEN10"] = cleaned_df["PDEN10"].map(PDEN10_mapping)
```

```python
cleaned_df['PDEN10'].value_counts()
```

```python
cleaned_df.rename(columns={"PDEN10": "POPDEN"}, inplace=True)
```

Replaced PDEN10 values with string-based values for easier understanding. Changed name to POPDEN. 

NOTE: COUTYP4 is recoded based on the 2013 Rural/Urban Continuum Codes.

COUTYP4 Len : 1 COUNTY METRO/NONMETRO STATUS (2013 3-LEVEL)

1 = Large Metro

2 = Small Metro

3 = Nonmetro

```python
cleaned_df["COUTYP4"].value_counts()
```

```python
COUTYP4_mapping = {
    1: "Large Metro",
    2: "Small Metro",
    3: "Nonmetro"
}
cleaned_df["COUTYP4"] = cleaned_df["COUTYP4"].map(COUTYP4_mapping)
```

```python
cleaned_df.rename(columns={"COUTYP4": "METROSTAT"}, inplace=True)
```

Replaced PDEN10 values with string-based values for easier understanding. Changed name to METROSTAT.
